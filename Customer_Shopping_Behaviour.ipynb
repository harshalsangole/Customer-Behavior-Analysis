```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#Create a coloumn Age_group

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels =labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#Purchase Frequency Days - Create Column
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Annually': 365,
    'Quarterly': 90,          # <-- Ensure this key matches exactly (case-sensitive)
    # add other frequencies here (e.g., 'Monthly': 30, 'Bi-Weekly': 14)
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
import urllib.parse

# Step 1: Connect to PostgreSQL
username = "postgres"
password = urllib.parse.quote_plus("postgres@123")  # Safely encodes '@' to '%40'
host = "localhost"
port = "5432"
database = "customer_behavior"

# Also ensure categorical columns are strings to avoid dtype insertion errors
df['age_group'] = df['age_group'].astype(str)

engine = create_engine(f"postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into PostgreSQL
table_name = "customer"
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
